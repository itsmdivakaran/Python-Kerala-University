## 1. Variables

A **variable** is a name that stores a value — like a labelled box.

```python
age = 45        # put 45 into a box called age
```

**Rules for names**
- Use letters, numbers and `_` (no spaces): `patient_name` ✅ `patient name` ❌
- Cannot start with a number: `age1` ✅ `1age` ❌
- Names are case-sensitive: `Age` and `age` are different.

````python
patient_name = "Anu"
age = 45
temperature = 37.8
is_admitted = True

print(patient_name)
print(age)
print(temperature)
print(is_admitted)
````

### 1.1 Data Types

Every value has a **type**. Use `type()` to check it.

| Type | Meaning | Example |
|------|---------|---------|
| `int` | whole number | `45` |
| `float` | decimal number | `37.8` |
| `str` | text (string) | `"Anu"` |

> 👀 **What to observe:** You see `str`, `int`, `float` and `bool` — one for each variable.

### 1.2 Changing a Variable

A variable can be given a new value at any time. The old value is replaced.
Python even allows a different type — this is called **dynamic typing**.

````python
age = 45
print("Age now:", age)

age = age + 1          # birthday!
print("Age after one year:", age)

age = "forty-six"      # now it is text (allowed, but be careful!)
print("Age as text:", age, type(age))

age = 46               # set it back to a number for the rest of the notebook
````

> 👀 **What to observe:** `age` changed from `45` to `46`, then became text. Python allows this, so always keep track of what is inside your variables.

### 1.3 Type Casting (Converting Types)

Data typed by users or read from files usually arrives as **text**. Convert it before doing maths.

| Function | Converts to | Example |
|----------|-------------|---------|
| `int()` | whole number | `int("45")` → `45` |
| `float()` | decimal | `float("37.8")` → `37.8` |
| `str()` | text | `str(45)` → `"45"` |

````python
age_text = "45"                    # this is text, not a number
print(age_text + "5")              # joins text -> 455

age_number = int(age_text)         # convert text to a number
print(age_number + 5)              # real maths -> 50

weight_text = "68.5"
weight = float(weight_text)
print(weight + 1.5)                # 70.0

print("Age is " + str(age_number)) # number to text, so we can join it
````

> 👀 **What to observe:** `'45' + '5'` gives `'455'` (joined text). After `int()`, `45 + 5` gives `50`.

### 1.4 f-strings — Printing Nicely

Put `f` before the quotes and write variables inside `{ }`.
Add `:.1f` to show 1 decimal place.

````python
name = "Anu"
age = 45
weight = 68.5
height = 1.62

bmi = weight / (height * height)

print(f"Patient {name} is {age} years old.")
print(f"BMI without rounding: {bmi}")
print(f"BMI rounded: {bmi:.1f}")
````

> 👀 **What to observe:** The f-string fills in the values for you. `:.1f` turns `26.10...` into `26.1`.

## 2. Storing Many Values Together

| Type | Looks like | Can change? | Allows duplicates? | Use it for |
|------|-----------|-------------|--------------------|-----------|
| **List** | `[37.2, 38.1]` | ✅ Yes | ✅ Yes | a series of readings |
| **Tuple** | `("Anu", 45)` | ❌ No | ✅ Yes | fixed details that should not change |
| **Set** | `{"Kochi", "Pune"}` | ✅ Yes | ❌ No | unique values |
| **Dictionary** | `{"name": "Anu"}` | ✅ Yes | keys are unique | a record with labels |

### 2.1 Lists

A **list** keeps values in order inside `[ ]`.
Positions (called **index**) start at **0**, not 1. `-1` means the last item.

```
temperatures = [36.8, 37.2, 38.1, 37.5, 36.9]
index:            0     1     2     3     4
```

````python
temperatures = [36.8, 37.2, 38.1, 37.5, 36.9]

print("All readings :", temperatures)
print("First reading:", temperatures[0])
print("Third reading:", temperatures[2])
print("Last reading :", temperatures[-1])
print("How many     :", len(temperatures))
print("Highest      :", max(temperatures))
print("Lowest       :", min(temperatures))
````

> 👀 **What to observe:** `temperatures[0]` is the **first** item. `len()` counts items; `max()` and `min()` find the highest and lowest.

### 2.2 Slicing a List

`list[start:stop]` gives a part of the list. The **stop position is not included**.

````python
temperatures = [36.8, 37.2, 38.1, 37.5, 36.9]

print(temperatures[0:3])   # positions 0, 1, 2
print(temperatures[2:])    # from position 2 to the end
print(temperatures[:2])    # from the start up to position 1
````

> 👀 **What to observe:** `[0:3]` gives 3 items (positions 0, 1, 2). Position 3 is not included.

### 2.3 Changing a List

| Method | What it does |
|--------|--------------|
| `.append(x)` | add `x` at the end |
| `.remove(x)` | remove the first `x` |
| `.sort()` | arrange in order |
| `list[i] = x` | replace the item at position `i` |

````python
patients = ["Anu", "Ravi", "Meera"]
print("Start      :", patients)

patients.append("John")        # new patient arrives
print("After add  :", patients)

patients.remove("Ravi")        # Ravi is discharged
print("After remove:", patients)

patients[0] = "Anu K."         # correct a name
print("After edit :", patients)

patients.sort()                # alphabetical order
print("Sorted     :", patients)
````

> 👀 **What to observe:** Lists can grow, shrink and change. This is what **mutable** means.

### 2.4 Tuples

A **tuple** uses `( )` and **cannot be changed** after it is created.
Use it for details that must stay fixed, like a patient's ID and date of birth.

````python
patient_info = ("P001", "Anu", "1980-05-12")

print("ID  :", patient_info[0])
print("Name:", patient_info[1])

# Unpacking: give each value its own name in one line
patient_id, name, dob = patient_info
print(f"{name} has ID {patient_id} and was born on {dob}")
````

````python
# Trying to change a tuple gives an error
patient_info = ("P001", "Anu", "1980-05-12")
try:
    patient_info[1] = "Ravi"
except TypeError as error:
    print("Error:", error)
````

> 👀 **What to observe:** Reading a tuple works like a list. Changing it gives a `TypeError` — that protection is the point of a tuple. (We learn `try/except` in Section 6.)

### 2.5 Sets

A **set** uses `{ }` and keeps only **unique** values. Duplicates are removed automatically.
Sets are also great for comparing two groups.

````python
visit_cities = ["Kochi", "Pune", "Kochi", "Delhi", "Pune"]
unique_cities = set(visit_cities)
print("Unique cities:", unique_cities)

morning_patients = {"Anu", "Ravi", "Meera"}
evening_patients = {"Ravi", "John"}

print("Came in both shifts  :", morning_patients & evening_patients)
print("Came in either shift :", morning_patients | evening_patients)
print("Only morning         :", morning_patients - evening_patients)
print("Is Anu in morning?   :", "Anu" in morning_patients)
````

> 👀 **What to observe:** The duplicates of Kochi and Pune disappear. `&` = in both, `|` = in either, `-` = in the first only. (The order of a set may look different on your screen — sets have no fixed order.)

### 2.6 Dictionaries

A **dictionary** stores **key : value** pairs inside `{ }`. Think of a patient form: the label is the key, the answer is the value.

```python
patient = {"name": "Anu", "age": 45}
patient["name"]   # -> "Anu"
```

````python
patient = {
    "id": "P001",
    "name": "Anu",
    "age": 45,
    "city": "Kochi",
    "temperature": 37.8,
}

print("Name:", patient["name"])
print("Age :", patient["age"])

patient["age"] = 46                 # update a value
patient["blood_group"] = "O+"       # add a new key
print(patient)

# .get() is safe: it does not crash if the key is missing
print("Phone:", patient.get("phone", "not available"))
````

> 👀 **What to observe:** Use the **key** in `[ ]` to read a value. `.get()` gives a default message instead of an error when the key does not exist.

````python
# Looping through a dictionary
patient = {"id": "P001", "name": "Anu", "age": 46, "city": "Kochi"}

for key, value in patient.items():
    print(key, "->", value)
````

> 👀 **What to observe:** `.items()` gives each **key** and its **value** together, one pair per line.

## 3. Making Decisions: `if`, `elif`, `else`

Python checks conditions from **top to bottom** and runs the **first** block that is `True`.

```python
if condition:
    do this
elif another_condition:
    do this instead
else:
    do this if nothing above was True
```

⚠️ The lines inside each block must be **indented** (4 spaces). Don't forget the colon `:`.

**Comparison operators**

| Symbol | Meaning |
|--------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `>` / `<` | greater / less than |
| `>=` / `<=` | greater or equal / less or equal |

````python
temperature = 38.4

if temperature >= 38.0:
    print("Fever - inform the doctor")
else:
    print("Normal temperature")
````

> 👀 **What to observe:** Since 38.4 is ≥ 38.0, the first message prints. Change `temperature` to `37.0` and run again.

````python
age = 70

if age < 18:
    print("Child")
elif age < 60:
    print("Adult")
else:
    print("Senior citizen")
````

> 👀 **What to observe:** Age 70 is not < 18 and not < 60, so the `else` block runs. Try `age = 10` and `age = 35`.

````python
# Blood pressure check (systolic value in mmHg)
systolic_bp = 135

if systolic_bp >= 140:
    print("High blood pressure")
elif systolic_bp >= 120:
    print("Elevated - monitor regularly")
else:
    print("Normal")
````

> 👀 **What to observe:** 135 is not ≥ 140, but it is ≥ 120, so we see **Elevated**. Order matters: always check the highest range first.

### 3.1 Combining Conditions: `and`, `or`, `not`

| Word | Meaning | Example |
|------|---------|---------|
| `and` | both must be True | `age >= 18 and age <= 60` |
| `or` | at least one True | `fever or cough` |
| `not` | reverses True/False | `not is_smoker` |

````python
age = 45
has_diabetes = False

# Eligible for the study: age between 18 and 60, and no diabetes
if age >= 18 and age <= 60 and not has_diabetes:
    print("Eligible for the study")
else:
    print("Not eligible")
````

````python
fever = True
cough = False

if fever or cough:
    print("Send to the flu clinic")
else:
    print("Send to the general clinic")
````

> 👀 **What to observe:** With `and`, every condition must be True. With `or`, one True is enough.

## 4. Loops: Repeating Work

### 4.1 `for` Loop

A `for` loop runs the same code **once for each item** in a list.

```python
for item in my_list:
    do something with item
```

````python
patients = ["Anu", "Ravi", "Meera", "John"]

for name in patients:
    print("Calling patient:", name)
````

> 👀 **What to observe:** The indented line runs 4 times, once for each name.

````python
# Loop + if together: check every temperature
temperatures = [36.8, 38.2, 37.1, 39.0, 36.5]

for temp in temperatures:
    if temp >= 38.0:
        print(temp, "-> Fever")
    else:
        print(temp, "-> Normal")
````

> 👀 **What to observe:** Each reading is checked one by one. Two readings show **Fever**.

````python
# Counting and adding inside a loop
temperatures = [36.8, 38.2, 37.1, 39.0, 36.5]

fever_count = 0
total = 0

for temp in temperatures:
    total = total + temp
    if temp >= 38.0:
        fever_count = fever_count + 1

average = total / len(temperatures)
print("Number of fever readings:", fever_count)
print(f"Average temperature: {average:.1f}")
````

> 👀 **What to observe:** We start the counters at `0` *before* the loop and update them *inside* the loop.

### 4.2 `range()` — Loop a Fixed Number of Times

`range(1, 6)` gives the numbers 1, 2, 3, 4, 5 (the end number is **not** included).

````python
for day in range(1, 6):
    print("Day", day, "- give medicine")
````

> 👀 **What to observe:** The loop runs 5 times: day 1 to day 5.

### 4.3 `enumerate()` and `zip()`

- `enumerate()` gives a **number** and the **item** together.
- `zip()` walks through **two lists side by side**.

````python
patients = ["Anu", "Ravi", "Meera"]

for number, name in enumerate(patients, start=1):
    print(f"Token {number}: {name}")
````

````python
names = ["Anu", "Ravi", "Meera"]
ages = [45, 32, 67]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
````

> 👀 **What to observe:** `enumerate` saves you from keeping your own counter. `zip` pairs the 1st name with the 1st age, the 2nd with the 2nd, and so on.

### 4.4 `while` Loop

A `while` loop repeats **as long as** a condition is True.
⚠️ Make sure the condition becomes False at some point, or the loop never stops.

````python
temperature = 39.5

while temperature >= 38.0:
    print(f"Temperature {temperature} - give medicine and check again")
    temperature = temperature - 0.5      # temperature goes down each hour

print(f"Temperature {temperature} - fever is gone")
````

> 👀 **What to observe:** The loop keeps running while the temperature is 38.0 or higher. It stops once it drops below 38.0.

### 4.5 `break` and `continue`

- `break` → **stop** the loop completely.
- `continue` → **skip** this item and go to the next one.

````python
# break: stop as soon as we find the first fever
temperatures = [36.8, 37.2, 38.5, 37.0, 39.1]

for temp in temperatures:
    if temp >= 38.0:
        print("First fever found:", temp)
        break
    print("Checked:", temp)
````

````python
# continue: skip readings that were not recorded (0 means missing)
temperatures = [36.8, 0, 37.5, 0, 38.2]

for temp in temperatures:
    if temp == 0:
        continue
    print("Valid reading:", temp)
````

> 👀 **What to observe:** With `break` the loop ends at 38.5 — the last values are never checked. With `continue` the zeros are skipped but the loop goes on.

## 5. Comprehensions — Short One-Line Loops

A **list comprehension** builds a new list in one line.

```python
new_list = [what_to_keep  for item in old_list  if condition]
```

Compare the two ways below — they give the **same result**.

````python
temps_f = [98.6, 100.4, 99.1, 102.2]      # Fahrenheit

# Normal loop
temps_c = []
for f in temps_f:
    temps_c.append(round((f - 32) * 5 / 9, 1))
print("Using a loop         :", temps_c)

# List comprehension (same thing in one line)
temps_c = [round((f - 32) * 5 / 9, 1) for f in temps_f]
print("Using a comprehension:", temps_c)
````

````python
temperatures = [36.8, 38.2, 37.1, 39.0, 36.5]

# Keep only the fever readings
fevers = [t for t in temperatures if t >= 38.0]
print("Fever readings:", fevers)

# Label every reading
labels = ["Fever" if t >= 38.0 else "Normal" for t in temperatures]
print("Labels:", labels)
````

````python
# Dictionary comprehension: name -> age in years after 5 years
ages = {"Anu": 45, "Ravi": 32, "Meera": 67}

ages_in_5_years = {name: age + 5 for name, age in ages.items()}
print(ages_in_5_years)
````

> 👀 **What to observe:** A comprehension is just a shorter loop. If it becomes hard to read, a normal `for` loop is perfectly fine.

## 6. Functions — Reusable Code

A **function** is a named block of code you can use again and again.
Use `def` to create it and `return` to send back the answer.

```python
def function_name(input1, input2):
    answer = ...
    return answer
```

**Why?** Write the BMI formula once, then use it for 100 patients.

````python
def calculate_bmi(weight, height):
    bmi = weight / (height * height)
    return round(bmi, 1)

print(calculate_bmi(68.5, 1.62))
print(calculate_bmi(82.0, 1.75))
print(calculate_bmi(55.0, 1.58))
````

> 👀 **What to observe:** One function, three patients, three answers — no copy-paste.

````python
# A function can contain if / elif / else
def fever_status(temperature):
    if temperature >= 39.0:
        return "High fever"
    elif temperature >= 38.0:
        return "Fever"
    else:
        return "Normal"

for temp in [36.9, 38.4, 39.6]:
    print(temp, "->", fever_status(temp))
````

> 👀 **What to observe:** The function decides the label; the loop just calls it for each reading.

### 6.1 Default Values and Keyword Arguments

- A **default value** is used when you don't pass that input.
- **Keyword arguments** let you name inputs when calling: `greet(name="Anu")`.

````python
def greet_patient(name, hospital="City Hospital"):
    return f"Welcome {name} to {hospital}!"

print(greet_patient("Anu"))                          # uses the default hospital
print(greet_patient("Ravi", "Medical College"))     # gives a new hospital
print(greet_patient(hospital="Rural Clinic", name="Meera"))   # keyword arguments
````

> 👀 **What to observe:** If `hospital` is not given, Python uses `City Hospital`. With keyword names, the order doesn't matter.

### 6.2 `*args` and `**kwargs`

- `*args` → accept **any number** of values (they arrive as a tuple).
- `**kwargs` → accept **any number of named** values (they arrive as a dictionary).

````python
def average_reading(*readings):
    return round(sum(readings) / len(readings), 1)

print(average_reading(120, 130))
print(average_reading(120, 130, 125, 128))
````

````python
def patient_card(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

patient_card(name="Anu", age=45, city="Kochi")
print("---")
patient_card(name="Ravi", blood_group="B+")
````

> 👀 **What to observe:** `average_reading` works with 2 values or 4 values. `patient_card` accepts whatever details you pass.

### 6.3 `lambda`, `map()` and `filter()`

A **lambda** is a tiny one-line function without a name: `lambda x: x * 2`.

- `map(function, list)` → apply the function to **every** item.
- `filter(function, list)` → **keep** only items where the function gives True.

````python
weights_kg = [68, 82, 55, 90]

weights_lb = list(map(lambda kg: round(kg * 2.2, 1), weights_kg))
print("Weights in pounds:", weights_lb)

heavy = list(filter(lambda kg: kg > 80, weights_kg))
print("Weights above 80 kg:", heavy)

# lambda is often used to sort
patients = [("Anu", 45), ("Ravi", 32), ("Meera", 67)]
by_age = sorted(patients, key=lambda p: p[1])
print("Sorted by age:", by_age)
````

> 👀 **What to observe:** `map` changed every weight; `filter` kept only two. `key=lambda p: p[1]` tells `sorted` to use the age.

## 7. Handling Errors: `try` and `except`

Real data has mistakes like `"N/A"` or empty values. Without protection, one bad value stops the whole program.

```python
try:
    risky code
except ErrorType:
    what to do if it fails
finally:
    always runs (success or failure)
```

````python
# Without try/except this would crash
text = "N/A"
try:
    value = float(text)
    print("Converted:", value)
except ValueError:
    print(f"Could not convert {text!r} to a number")
````

````python
# Cleaning a list of temperatures typed by different nurses
raw_readings = ["36.8", "37.5", "N/A", "38.2", "", "high", "37.0"]

clean_readings = []
bad_readings = []

for item in raw_readings:
    try:
        clean_readings.append(float(item))
    except ValueError:
        bad_readings.append(item)

print("Clean readings:", clean_readings)
print("Bad readings  :", bad_readings)
````

> 👀 **What to observe:** The program does not crash. Good values are kept, bad ones are collected so we can report them.

````python
# finally: always runs
total_patients = 0

try:
    average = 500 / total_patients
except ZeroDivisionError:
    print("No patients yet - cannot calculate the average")
finally:
    print("Report check finished")
````

> 👀 **What to observe:** Dividing by zero gives an error, `except` handles it, and `finally` still runs.

## 8. Using Modules (Ready-Made Code)

A **module** is a file of Python code that someone already wrote. Use `import` to bring it in.

## 9. Mini Project — Daily Clinic Report

Let's put everything together: a list of dictionaries, a loop, `if/else`, a function and an f-string.

````python
clinic_patients = [
    {"name": "Anu",   "age": 45, "temperature": 37.2},
    {"name": "Ravi",  "age": 32, "temperature": 38.6},
    {"name": "Meera", "age": 67, "temperature": 36.9},
    {"name": "John",  "age": 58, "temperature": 39.2},
]

def fever_status(temperature):
    if temperature >= 38.0:
        return "Fever"
    return "Normal"

fever_count = 0

print("Name     Age   Temp   Status")
print("-" * 30)
for p in clinic_patients:
    status = fever_status(p["temperature"])
    if status == "Fever":
        fever_count += 1
    print(f"{p['name']:<8} {p['age']:<5} {p['temperature']:<6} {status}")

print("-" * 30)
print(f"Total patients: {len(clinic_patients)} | With fever: {fever_count}")
````

> 👀 **What to observe:** A neat little report made with only basic Python. In Session 3, Pandas will do this kind of work in one or two lines.

````python
#@title Solution 2 — Blood pressure list
bp_readings = [118, 145, 132, 160, 110, 128]

high_count = 0
for bp in bp_readings:
    if bp >= 140:
        print(bp, "-> High")
        high_count += 1
    else:
        print(bp, "-> OK")

print("Number of high readings:", high_count)

high_readings = [bp for bp in bp_readings if bp >= 140]
print("High readings:", high_readings)
````

````python
#@title Solution 3 — Patient checker function
patients = [
    {"name": "Anu",   "temperature": "37.4"},
    {"name": "Ravi",  "temperature": "38.9"},
    {"name": "Meera", "temperature": "N/A"},
    {"name": "John",  "temperature": "39.5"},
]

def check_patient(patient):
    try:
        temp = float(patient["temperature"])
    except ValueError:
        return "Check again"
    if temp >= 38.0:
        return "Fever"
    return "Normal"

for p in patients:
    print(f"{p['name']}: {check_patient(p)}")
````
